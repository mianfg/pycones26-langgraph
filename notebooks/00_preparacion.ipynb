## 5 · ¿Se dibujan los grafos?

Deberías ver un diagrama con tres cajas: `__start__ → saludar → __end__`, y debajo el resultado `{'nombre': '¡Hola, PyConES!'}`.
Si ves texto en vez de un dibujo, usa JupyterLab (`uv run jupyter lab`) o cambia a
`mostrar_grafo(grafo, png=True)` (necesita internet).

````python
from typing import TypedDict

from langgraph.graph import END, START, StateGraph

from taller.viz import mostrar_grafo


class Estado(TypedDict):
    nombre: str


def saludar(state: Estado) -> dict:
    return {"nombre": f"¡Hola, {state['nombre']}!"}


grafo = StateGraph(Estado).add_node(saludar).add_edge(START, "saludar").add_edge("saludar", END).compile()
mostrar_grafo(grafo)
grafo.invoke({"nombre": "PyConES"})
````
